usando el teorema 3 (matriz P(t)):

aproximar P(t) usando los primeros M términos de la serie infinita. Se obtienen buenos resultados si se elije:
$$M=max\{rt+5\sqrt{rt},20\}$$

1. Use esta propuesta para calcular P(0.5), P(1) y P(5) para la matriz R del ejercicio 1.

```python
#imprtar las librerias necesarias

import sympy as sp
import math
import numpy as np
```

```python
R=sp.Matrix([[0,2,3,0],[4,0,2,0],[0,2,0,2],[1,0,3,0]])
```

```python
rs = [sum(fila) for fila in R.tolist()]
print('ri=',rs)
r=max(rs)
print('r=',r)
```

```python
#contruccion de la matriz P gorrito
def pg(R,r):
  n=R.rows #definir la dimension de la matriz
  P=sp.zeros(n,n) #hacer una matriz del mismo tamaño de R
  for i in range(n):#recorre columnas
    ri=np.sum(R[i,:])#sumar las filas
    for j in range(n): #recorre filas
      if i==j:
        P[i,j]=1-ri/r
      else:
        P[i,j]=R[i,j]/r
  return P

PG=pg(R,r)
sp.pprint(PG.evalf(2))
```

escribir P(t) usando el teorema para el ejemplo y calcular P(0.5),P(1),P(5)

```python
#funcion para contruir la matriz del teorema 3
def P(t,r,Pg):
  n=Pg.rows #definir la dimension de la matriz
  P=sp.zeros(n,n) #hacer una matriz del mismo tamaño de P gorro
  k= max((r*t)+(5*math.sqrt(r*t)),20) #encontrar el M optimo
  for i in range(int(k)+1):
    P=P+(math.exp(-r*t)*((r*t)**i/math.factorial(i))*Pg**i)#contruitr la matriZ P
  return P

#definir los  tiempos que se pieden en el ejercicio
t1=0.5
t2=1
t3=5
#obtener las matrices
P1=P(r,t1,Pg)
P2=P(r,t2,Pg)
P3=P(r,t3,Pg)
#impresion de los resultados
print('P(0.5)=\n')
sp.pprint(P1.evalf(6))
print('\n P(1)=\n')
sp.pprint(P2.evalf(6))
print('\n P(5)=\n')
sp.pprint(P3.evalf(6))
```

Teorema 4 (cotas de error para P(t))

$$|p_{ij}(t)-p_{ij}^M(t)|\leq \sum_{k=M+1}^\infty e^{-rt}\frac{(rt)^k}{k!} \qquad \qquad \forall \quad 1\leq i,j\leq N$$
Este teorema se puede usar así: Suponga que se desea calcular P(t)
con una tolerancia $\varepsilon$. Elija M tal que:
$$\sum_{k=M+1}^\infty e^{-rt}\frac{(rt)^k}{k!}\leq \varepsilon$$

```python
def uniformar(R,t,eps):
  n=Pg.rows #definir la dimension de la matriz
  PG=sp.zeros(n,n) #hacer una matriz del mismo tamaño de R
  r=max(np.sum(R,axis=1))#obetener el maximo de los ri

  #construccion de P gorrito
  for i in range(n):#recorre columnas
    ri=np.sum(R[i,:])#sumar las filas para obtener los ri
    for j in range(n): #recorre filas
      if i==j:
        PG[i,j]=1-ri/r
      else:
        PG[i,j]=R[i,j]/r

  #Algoritmo
  A=PG.copy()
  rt=r*t
  B=math.exp(-rt)*sp.eye(n)
  c=math.exp(-rt)
  suma=c
  k=1
  while suma <1-eps:
    c=c*(rt)/k
    B=B+c*A
    A=A*PG
    suma=suma+c
    k=k+1
  print(f'termina en k={k-1} iteracines')
  print(f'suma final: {suma:.10f}')
  print(f'error max. estimado={1-suma:.2e}')
  return B,k-1

eps=0.00001
resultados={}

for t in [0.5,1,5]:
  Pt,Mfinal=uniformar(R,t,eps)
  resultados[t]=Pt
  print(f'\nP({t}) calculada:\n')
  sp.pprint(Pt.evalf(6))
  print('\n\n')
```
